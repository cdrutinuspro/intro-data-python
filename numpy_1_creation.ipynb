## 1. Pourquoi NumPy ?

Une liste Python peut contenir n'importe quoi, ce qui la rend flexible mais lente. Une `array` NumPy contient des éléments **tous du même type**, stockés de façon compacte, et permet des calculs **vectorisés** (sans boucle Python).

````python
import numpy as np
import time

N = 2_000_000
liste = list(range(N))
arr = np.arange(N)

t0 = time.perf_counter()
res_liste = [x * 2 for x in liste]
t1 = time.perf_counter()
res_arr = arr * 2
t2 = time.perf_counter()

print(f"Liste : {t1 - t0:.4f} s")
print(f"NumPy : {t2 - t1:.4f} s")
````

> NumPy est typiquement 10 à 100 fois plus rapide sur ce genre d'opération.

## 2. Créer une array à partir d'une liste

La méthode « classique » : `np.array(...)`.

````python
a = np.array([1, 2, 3, 4])
print(a)
print(type(a))
````

Avec une liste de listes, on obtient une array à **2 dimensions** (une matrice).

````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m)
````

### Les attributs importants

| Attribut | Signification |
|---|---|
| `shape` | dimensions (lignes, colonnes, …) |
| `ndim` | nombre de dimensions |
| `size` | nombre total d'éléments |
| `dtype` | type des éléments |

````python
print("shape :", m.shape)
print("ndim  :", m.ndim)
print("size  :", m.size)
print("dtype :", m.dtype)
````

### Le type (`dtype`)

Tous les éléments partagent le même type. NumPy **convertit** si nécessaire :

````python
print(np.array([1, 2, 3]).dtype)          # entiers
print(np.array([1, 2.5, 3]).dtype)        # un float => tout devient float
print(np.array([1, 2, 3], dtype=float))   # on peut forcer le type
````

## 3. Arrays « pré-remplies »

Quand on connaît la forme mais pas encore les valeurs :

````python
print(np.zeros(5))
print()
print(np.ones((2, 3)))      # attention : on passe un tuple (lignes, colonnes)
print()
print(np.full((2, 2), 7))
print()
print(np.eye(3))            # matrice identité
````

## 4. `arange` : une suite avec un **pas**

`np.arange(début, fin, pas)` ressemble à `range`, mais renvoie une array (et accepte les flottants). La borne de fin est **exclue**.

````python
print(np.arange(5))            # 0 à 4
print(np.arange(2, 10))         # 2 à 9
print(np.arange(0, 10, 2))      # pas de 2
print(np.arange(1, 0, -0.25))   # pas négatif
````

## 5. `linspace` : une suite avec un **nombre de points**

`np.linspace(début, fin, nb_points)` répartit régulièrement `nb_points` valeurs, **bornes incluses** (par défaut). C'est l'outil de base pour tracer une fonction.

````python
print(np.linspace(0, 1, 5))
print(np.linspace(0, 10, 3))
````

### `arange` ou `linspace` ?

- Je connais le **pas** → `arange`
- Je connais le **nombre de points** et je veux inclure la borne de fin → `linspace`

⚠️ Avec des flottants, `arange` peut surprendre sur le nombre d'éléments (erreurs d'arrondi). Comparez :

````python
print(len(np.arange(0, 1, 0.1)))
print(len(np.linspace(0, 1, 10, endpoint=False)))
print(np.arange(0, 1.0001, 0.1))
````

## 6. Arrays aléatoires

On crée un générateur (avec une *graine* pour avoir des résultats reproductibles), puis on tire des valeurs.

````python
rng = np.random.default_rng(seed=42)

print(rng.random(3))                 # 3 flottants dans [0, 1[
print(rng.integers(1, 7, size=5))    # 5 lancers de dé (1 à 6)
print(rng.normal(0, 1, size=(2, 3))) # loi normale (moyenne 0, écart-type 1)
````

## 7. Changer la forme : `reshape`

On peut réorganiser les mêmes données en une autre forme (le nombre total d'éléments doit rester le même).

````python
a = np.arange(12)
print(a.reshape(3, 4))
print()
print(a.reshape(2, -1))   # -1 : NumPy calcule la dimension manquante
````

## Récapitulatif

| Je veux… | Fonction |
|---|---|
| depuis une liste | `np.array([...])` |
| remplie de 0 / 1 / constante | `np.zeros`, `np.ones`, `np.full` |
| suite avec un pas | `np.arange(début, fin, pas)` |
| suite avec un nombre de points | `np.linspace(début, fin, n)` |
| valeurs aléatoires | `rng = np.random.default_rng(seed)` |
| changer la forme | `.reshape(...)` |
