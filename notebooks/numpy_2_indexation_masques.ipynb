## 1. Indexation d'une array 1D

Comme pour les listes : on compte à partir de **0**, et les indices négatifs partent de la fin.

````python
a = np.array([10, 20, 30, 40, 50])
print(a[0])    # premier
print(a[2])
print(a[-1])   # dernier
print(a[-2])   # avant-dernier
````

## 2. Slicing 1D : `a[début:fin:pas]`

La borne de **fin est exclue**. Chaque borne est optionnelle.

````python
a = np.arange(10)   # [0 1 2 ... 9]
print(a[2:5])    # indices 2, 3, 4
print(a[:3])     # du début à l'indice 2
print(a[7:])     # de l'indice 7 à la fin
print(a[::2])    # un élément sur deux
print(a[::-1])   # array inversée
````

## 3. Indexation 2D : `m[ligne, colonne]`

On sépare les indices par une virgule (et non `m[i][j]`, qui marche mais est moins efficace).

````python
m = np.arange(1, 13).reshape(3, 4)
print(m)
print()
print(m[0, 0])    # ligne 0, colonne 0
print(m[1, 2])    # ligne 1, colonne 2
print(m[-1, -1])  # dernier élément
````

### Slicing 2D

On peut « trancher » chaque axe séparément. Le caractère `:` seul signifie « tout cet axe ».

````python
print(m[0, :])     # ligne 0 entière
print(m[:, 1])     # colonne 1 entière
print(m[0:2, 1:3]) # sous-matrice (lignes 0-1, colonnes 1-2)
print(m[:, ::2])   # une colonne sur deux
````

### ⚠️ Un slice est une *vue*, pas une copie

Modifier un slice modifie l'array d'origine ! Utilisez `.copy()` pour obtenir une copie indépendante.

````python
a = np.arange(5)
vue = a[1:4]
vue[0] = 99
print(a)           # a a été modifiée

b = np.arange(5)
copie = b[1:4].copy()
copie[0] = 99
print(b)           # b est intacte
````

### Indexation par liste d'indices (*fancy indexing*)

On peut aussi fournir une liste d'indices pour piocher des éléments précis (le résultat est cette fois une **copie**).

````python
a = np.array([10, 20, 30, 40, 50])
print(a[[0, 2, 4]])
print(m[[0, 2], :])   # lignes 0 et 2
````

## 4. Tests et masques booléens

Une comparaison appliquée à une array renvoie une **array de booléens** de même forme, appelée *masque*.

````python
notes = np.array([12, 7, 15, 9, 18, 4, 11])
masque = notes >= 10
print(masque)
print(masque.dtype)
````

On utilise ensuite ce masque comme un **indice** : seuls les éléments où le masque vaut `True` sont conservés.

````python
print(notes[masque])
print(notes[notes >= 10])   # en une seule ligne
````

### Combiner plusieurs conditions

On utilise `&` (et), `|` (ou), `~` (non), **avec des parenthèses** autour de chaque condition. Les mots `and`, `or`, `not` ne fonctionnent pas sur des arrays.

````python
print(notes[(notes >= 10) & (notes <= 15)])
print(notes[(notes < 8) | (notes > 16)])
print(notes[~(notes >= 10)])
````

### Compter et vérifier

Un booléen vaut 1 (`True`) ou 0 (`False`) : on peut donc **compter** avec `sum`.

````python
print("Nombre de notes >= 10 :", (notes >= 10).sum())
print("Proportion de reçus     :", (notes >= 10).mean())
print("Au moins une note > 17 ?", (notes > 17).any())
print("Toutes >= 10 ?          ", (notes >= 10).all())
````

### Modifier avec un masque

On peut aussi **affecter** des valeurs aux éléments sélectionnés.

````python
n2 = notes.copy()
n2[n2 < 10] = 10     # on remonte les notes basses à 10
print(n2)
````

### `np.where` : choisir selon une condition

`np.where(condition, si_vrai, si_faux)` construit une nouvelle array. Avec la condition seule, il renvoie les **indices** des éléments `True`.

````python
print(np.where(notes >= 10, "reçu", "ajourné"))
print(np.where(notes >= 10))   # indices des notes >= 10
````

### Masques en 2D

Ça marche pareil : le masque a la même forme que la matrice, mais le résultat filtré est « aplati » en 1D.

````python
m = np.arange(1, 13).reshape(3, 4)
print(m[m % 2 == 0])     # tous les éléments pairs
m[m > 8] = 0
print(m)
````

## Récapitulatif

| Je veux… | Syntaxe |
|---|---|
| un élément | `a[i]`, `m[i, j]` |
| une tranche | `a[d:f:p]`, `m[l1:l2, c1:c2]` |
| une copie indépendante | `.copy()` |
| filtrer selon une condition | `a[a > 3]` |
| combiner des conditions | `(c1) & (c2)`, `(c1) \| (c2)`, `~(c)` |
| compter / vérifier | `.sum()`, `.any()`, `.all()` |
| remplacer selon une condition | `np.where(cond, a, b)` |
